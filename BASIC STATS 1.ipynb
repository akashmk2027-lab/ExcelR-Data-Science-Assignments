# Data Visualization

## Objective: To visualize the distribution and relationship of numerical and categorical variables in the dataset.

### Histograms

#### Histograms are used to visualize the distribution of numerical variables in the dataset. They show the frequency of observations within different ranges of values. Histograms help in understanding the shape of the data distribution and identifying possible skewness and unusual values.

```python
df['Volume'].plot(kind = 'hist',bins = 40)
plt.xlabel('Volume')
plt.ylabel('Frequency')
plt.title('Distribution of Volume')
plt.show()
```

### Interpretation of Volume Histogram
#### The histogram shows that the Volume values are mainly concentrated between 2 and 6 units, with the highest frequency occurring around the lower volume range.

#### The distribution is **right-skewed (positively skewed)** because the histogram has a long tail extending toward higher volume values. A small number of observations have much higher volume values, reaching up to approximately 31 units. These observations may be considered potential extreme values.

#### Overall, most observations have relatively low volumes, while only a small number of observations have high volumes.

```python
df['Avg Price'].plot(kind = 'hist',bins = 40 )
plt.xlabel('Avg Price')
plt.ylabel('Frequency')
plt.title('Distribution of Avg Price')
plt.show()
```

### Interpretation of Avg Price Histogram

#### The histogram shows that most of the **Avg Price** observations are concentrated at relatively lower values, while only a smaller number of observations occur at much higher prices.

#### The distribution is **strongly right-skewed (positively skewed)**, as there is a long tail extending toward higher values, reaching approximately 60,100.

#### There are also several observations at much higher price levels compared with the majority of the data. These may represent **potential extreme values or high-priced products**.

#### Overall, the Avg Price variable has a highly uneven distribution, with most observations at lower prices and a relatively small number of observations at substantially higher prices.

# Box Plots
### Box plots are used to visualize the distribution of numerical variables and identify potential outliers. They display the median, first quartile (Q1), third quartile (Q3), and the interquartile range (IQR).

### The interquartile range is calculated as:

#### **IQR = Q3 - Q1**

### Values falling below Q1 − 1.5 × IQR or above Q3 + 1.5 × IQR are identified as potential outliers. Box plots also help in comparing the spread and distribution of the numerical variables.

```python
# Box plot of Volume
df['Volume'].plot(kind = 'box')
plt.show()
```

### Interpretation of Volume Box Plot
#### The box plot shows that the **Volume** variable has a median of approximately 4 units. The middle 50% of the observations lie roughly between 3 and 6 units, representing the interquartile range (IQR).

#### The upper whisker extends to approximately 10 units. Several individual points are visible above the upper whisker, indicating **potential outliers**. Some of these values are considerably higher than the majority of observations, with the highest value reaching approximately 31 units.

#### Overall, the box plot indicates that Volume has a **right-skewed distribution** with several high-volume observations that may be considered extreme values.

```python
# Box plot of Avg Price
df['Avg Price'].plot(kind = 'box')
plt.show()
```

#### Overall, the box plot shows that Avg Price has a **strongly right-skewed distribution**, with substantial variation and several high-value observations.

### Identify Data Types

#### The data types of all columns in the dataset are identified using the df.dtypes function. This helps distinguish between numerical and categorical variables.

#### Categorical columns (object): Date, Day, SKU, City, BU, Brand, and Model.
