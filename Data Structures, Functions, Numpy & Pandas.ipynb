### 11.	Create a small DataFrame manually with 10 rows and columns: 'Name', 'Age', 'City', and 'Salary'. Then:
- Use .info(), .describe()
- Select 'Name' and 'City' columns
- Drop 'City' column
- Fill any missing values in 'Salary' column with the mean
- Remove any duplicate rows

```python
import pandas as pd
import numpy as np
```

```python
# Create a small DataFrame with 10 rows
# np.nan represents a missing value in the Salary column
dic1 = {'Name': ['Akash', 'Rahul', 'Anu', 'Arun', 'Meera','Vijay', 'Anu', 'Riya', 'Kiran', 'Rahul'],
        'Age': [23, 25, 22, 28, 24, 30, 22, 26, 27, 25], 
        'City': ['Kannur', 'Kochi', 'Calicut', 'Kannur', 'Kochi','Delhi', 'Calicut', 'Mumbai', 'Bangalore', 'Kochi'],
        'Salary': [25000, 30000, 28000,np.nan ,35000,40000, 28000, 32000, 45000, 30000]}

df = pd.DataFrame(dic1)
df
```

```python
# Display basic information about the DataFrame
df.info()
```

```python
# Display statistical summary of numerical columns
df.describe()
```

```python
# Select only the Name and City columns
df[['Name','City']]
```

```python
# Drop the City column
df = df.drop('City', axis=1)
df
```

```python
# Fill the missing Salary value with the mean salary
df['Salary']=df['Salary'].fillna(value= df['Salary'].mean())
df
```

```python
# Remove duplicate rows
df = df.drop_duplicates()
df
```

```python
# Display the final DataFrame
df
```
